## 13

```python
A=int(input())
n=0
def S(n):
    a=0
    for k in range(1,n+1):
        a+=1/k
    return a

while True:
    if S(n)<A:
        n+=1
    else:
        break

print(n)
```

## 16

```python
n=int(input())
count_even=0
count_odd=0
while True:
    check=n%10
    if check%2==0:
        count_even+=1
    else:
        count_odd+=1
    n=n//10
    if n==0:
        break

print(f"even counted: {count_even}, odd counted: {count_odd}")
```

```python
n=int(input())
count_element=0
def check(x):
    so=0
    for k in range(1,x+1):
        if x%k==0:
            so+=1
    if so ==2:
        return True
    else:
        return False
while True:
    a=n%10
    if check(a)==True :
        count_element+=1
    n=n//10
    if n==0:
        break

print(count_element)
```
