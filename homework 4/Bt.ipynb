# W4A21


```python
n=int(input())
def sum_digit(n):
    sm=0
    while True:
        a=n%10
        sm+=a
        n=n//10
        if n==0:
            break
    return sm
print(sum_digit(n))
```

# W4A22



```python
n=int(input())
count_even=0
count_odd=0
while True:
    check=n%10
    if check%2==0:
        count_even+=1
    else:
        count_odd+=1
    n=n//10
    if n==0:
        break

print(f"even counted: {count_even}, odd counted: {count_odd}")
```

# W4A23


```python
n=int(input())
k=0
def S(k):
    a=0
    for k in range(1,k+1):
        a+=k
    return a

while True:
    if S(k)<n:
        k+=1
    else:
        break

print(k-1)
```

# W4A24


```python
A=int(input())
n=0
def S(n):
    a=0
    for k in range(1,n+1):
        a+=1/k
    return a

while True:
    if S(n)<A:
        n+=1
    else:
        break

print(n)
```

# W4A25


```python
mx=-1
mn=-1
while True:
    a=int(input())
    if a==-1:
        break
    if a>mx or mx==-1:
        mx=a
    if a<mn or mn==-1:
        mn=a
print(f"Max: {mx}, Min: {mn}")
```

# W4A27



```python
a=input() + " "
check_string=False
count_letter=0
for k in a:
    if check_string==True and k!=" ":
        check_string=True
    elif check_string==True and k==" ":
        count_letter +=1
        check_string=False
    elif check_string ==False and k!=" ":
        check_string=True
    elif check_string== False and k==" ":
        check_string=False
print(count_letter)
```
