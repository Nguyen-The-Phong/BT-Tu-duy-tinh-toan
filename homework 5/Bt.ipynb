```python
quad=[]
for k in range(4):
    a=tuple(map(float,input().split()))
    quad.append(a)
p=tuple(map(float,input().split()))

def PT_DT(x,y):
    phap_tuyen=[-(y[1]-x[1]),y[0]-x[0]]
    he_so=0-(phap_tuyen[0]*x[0]+phap_tuyen[1]*x[1])
    phap_tuyen.append(he_so)
    return phap_tuyen

def check_same_side(dt,x,p):
    check=1
    for k in [x,p]:
        check*=dt[0]*k[0]+dt[1]*k[1]+dt[2]
    if check>=0:
        return True
    else:
        return False


def point_in_quadrilateral(quad, p):
    dt1=PT_DT(quad[0],quad[1])
    dt2=PT_DT(quad[1],quad[2])
    dt3=PT_DT(quad[2],quad[3])
    dt4=PT_DT(quad[3],quad[0])
    if check_same_side(dt1,quad[2],p) == False or check_same_side(dt2,quad[3],p)==False or check_same_side(dt3,quad[1],p)==False or check_same_side(dt4,quad[2],p)==False:
        return False
    else:
        return True
print(point_in_quadrilateral(quad,p))
```

```python
n=int(input())

def difference(x):
    if len(str(x))==len(set(str(x))):
        return True
    else:
        return False

def sum_difference(n):
    sm=0
    if n<=0:
        return 0
    for k in range(1,n+1):
        if difference(k)==True:
            sm+=k
    return sm
print(sum_difference(n))
```
