```python
import math
a,b,c = map(float,input().split())
def S(x,y,z):
    p=(x+y+z)/2
    return math.sqrt(p*(p-x)*(p-y)*(p-z))

def CV(x,y,z):
    return x+y+z

if a+b<=c or a+c<=b or b+c<=a:
    print("Khong la ba canh cua mot tam giac")
elif a==b==c:
    print("Đều")
    print(f"{S(a,b,c)} {CV(a,b,c)}")
elif a==b or a==c or b==c:
    print("Cân")
    print(f"{S(a,b,c)} {CV(a,b,c)}")
elif a**2 + b**2 == c**2 or c**2 + b**2 == a**2 or a**2 + c**2 == b**2:
    print("Vuông")
    print(f"{S(a,b,c)} {CV(a,b,c)}")
else: 
    print("Tam giác thường")
    print(f"{S(a,b,c)} {CV(a,b,c)}")

```

```python
import math
print("ax^2 + bx + c=0")
a=float(input("a= "))
b=float(input("b= "))
c=float(input("c= "))

if a==0:
    if b==0 and c ==0:
        print("PT vô số nghiệm")
    elif b==0:
        print("PT vô nghiệm")
    else:
        print(f"PT có nghiệm duy nhất: x={-c/b}")
else:
    delta=b**2-4*a*c
    if delta <0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        print(f"Phương trình có nghiêm kép: x={-b/(2*a)}")
    else:
        print(f"Phương trình có 2 nghiệm phân biệt x1 ={(-b-math.sqrt(delta))/(2*a)}, x2={(-b+math.sqrt(delta))/(2*a)}")
```

```python
a=int(input())
lst=[]
while a!=0:
    lst.append(a-(a//60)*60)
    a=a//60
lst.reverse()
while len(lst)<3:
    lst.insert(0,0)
while len(lst)>3:
    stole=lst.pop(0)*60
    lst[0]+=stole
lst=list(map(str,lst))
print(f"{lst[0]} giờ, {lst[1]} phút, {lst[2]} giây")
```

```python
x,y=map(float,input("nhập tọa độ tâm đường tròn I(x,y) (2 số cách nhau 1 dấu cách ): ").split())
r=float(input("nhập bán kính đường tròn: "))
a,b=map(float,input("nhập tọa A(x,y) (2 số cách nhau 1 dấu cách ): ").split())
def check(x,y,r,a,b):
    cyrene=(a-x)**2 + (b-y)**2 - r**2
    if cyrene == 0:
        return f"A({a},{b}) thuộc đường tròn tâm I({x},{y})"
    else:
        return f"A({a},{b}) không thuộc đường tròn tâm I({x},{y})"

print(check(x,y,r,a,b))
```
