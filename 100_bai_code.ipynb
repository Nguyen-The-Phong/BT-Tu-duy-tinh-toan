## 94


```python
a,b=map(int,input().split())
c,d=map(int,input().split())
def UCLN(x,y):
    A=[]
    for k in range(1,x+1):
        if x%k==0:
            A.append(k)
    A.reverse()
    for k in A:
        if y%k==0:
            return k
def plus(x,y,z,w):
    tu=x*w+y*z
    mau=y*w
    if tu>mau:
        he_so=tu//mau
        tu=tu-he_so*mau
        print(f"{he_so},{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")
    else:
        print(f"{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")

def minus(x,y,z,w):
    tu=x*w-y*z
    mau=y*w
    if tu>mau:
        he_so=tu//mau
        tu=tu-he_so*mau
        print(f"{he_so},{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")
    else:
        print(f"{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")

def time(x,y,z,w):
    tu=x*z
    mau=y*w
    if tu>mau:
        he_so=tu//mau
        tu=tu-he_so*mau
        print(f"{he_so},{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")
    else:
        print(f"{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")

def devide(x,y,z,w):
    tu=x*w
    mau=y*z
    if tu>mau:
        he_so=tu//mau
        tu=tu-he_so*mau
        print(f"{he_so},{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")
    else:
        print(f"{tu/UCLN(tu,mau):.0f}/{mau/UCLN(tu,mau):.0f}")
plus(a,b,c,d)
minus(a,b,c,d)
time(a,b,c,d)
devide(a,b,c,d)
```

```python
a=int(input())
b=list(map(int,input().split()))
def sum_even(x):
    sm=0
    for k in x:
        if k%2==0:
            sm+=k
    return sm
def count_even(x):
    sm=0
    for k in x:
        if k%2==0:
            sm+=1
    return sm
print(f"{sum_even(b)/count_even(b):.2f}")
```

```python
a=list(map(int,input().split()))
def count_element(x):
    element=0
    for k in x:
        lst=0
        for i in range(1,k+1):
            if k%i==0:
                lst+=1
        if lst==2:
            element+=1
    return element
print(count_element(a))
```

```python
a=list(map(int,input().split()))
def element(x):
    element=[]
    for k in x:
        lst=0
        for i in range(1,k+1):
            if k%i==0:
                lst+=1
        if lst==2:
            element.append(k)
    return element
lst=element(a)
print(f"{sum(k for k in lst)/len(lst):.2f}")
```

## 99


```python
import math
a=list(map(int,input().split()))
def check_cp(x):
    cp=[]
    for k in x:
        if k>=0:
            b=math.sqrt(k)
            if int(b)**2==k:
                cp.append(k)
    return cp
cp=check_cp(a)
print(f"{sum(k for k in cp)/len(cp):.1f}")
```

```python
a=int(input())
b=list(map(int,input().split()))
def max(x):
    m=x[0]
    M=[]
    for k in range(len(x)):
        if m<x[k]:
            m=x[k]
            M.clear()
            M.append(k)
        elif m==x[k]:
            M.append(k)
    M.insert(0,m)
    return M

def min(x):
    m=x[0]
    M=[]
    for k in range(len(x)):
        if m>x[k]:
            m=x[k]
            M.clear()
            M.append(k)
        elif m==x[k]:
            M.append(k)
    M.insert(0,m)
    return M
for k in max(b):
    print(k,end=" ")
print("")
for k in min(b):
    print(k,end=" ")
```
